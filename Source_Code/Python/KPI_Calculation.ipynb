```python
import pandas as pd
import numpy as np
import os

finalized_path = r"C:\Users\ABISHEK.000\OneDrive\Desktop\DA\Final\Cross_country_Technology_Adoption\Finalized"

# ── LOAD FINALIZED CSVs ───────────────────────────────────
print("Loading finalized tables...")

df_media = pd.read_csv(os.path.join(finalized_path, 'media_tech_usage_final.csv'), parse_dates=['record_date'])
df_app   = pd.read_csv(os.path.join(finalized_path, 'app_download_comparison_final.csv'), parse_dates=['record_date'])
df_ai    = pd.read_csv(os.path.join(finalized_path, 'ai_transformation_2026_final.csv'), parse_dates=['record_date'])
df_comp  = pd.read_csv(os.path.join(finalized_path, 'company_adaptation_final.csv'), parse_dates=['record_date'])

print(f"df_media : {df_media.shape}")
print(f"df_app   : {df_app.shape}")
print(f"df_ai    : {df_ai.shape}")
print(f"df_comp  : {df_comp.shape}")
print("\nAll tables loaded.")
```

```python
# Pltform was not Standardised in SQL Cleaning
platform_map = {
    'youtube': 'YouTube',
    'NETFLIX': 'Netflix',
    'tiktok': 'TikTok'
}
df_media['platform'] = df_media['platform'].replace(platform_map)
```

```python
# ════════════════════════════════════════════════════════
# PAGE 1 — AI ADOPTION KPIs
# Tables: df_ai + df_comp
# ════════════════════════════════════════════════════════
```

```python
print("=" * 55)
print("PAGE 1 — AI ADOPTION KPIs")
print("=" * 55)

# KPI 1: Avg AI Adoption Rate % by Country
kpi1 = df_ai.groupby('country')['ai_adoption_rate_pct'].mean().round(2).reset_index()
kpi1.columns = ['Country', 'Avg_AI_Adoption_Rate_Pct']
print("\nKPI 1 — Avg AI Adoption Rate % by Country:")
print(kpi1.to_string(index=False))

# KPI 2: Avg AI Adoption Rate % by Sector
kpi2 = df_ai.groupby('sector')['ai_adoption_rate_pct'].mean().round(2).reset_index()
kpi2.columns = ['Sector', 'Avg_AI_Adoption_Rate_Pct']
kpi2 = kpi2.sort_values('Avg_AI_Adoption_Rate_Pct', ascending=False)
print("\nKPI 2 — Avg AI Adoption Rate % by Sector:")
print(kpi2.to_string(index=False))

# KPI 3: AI Adoption Velocity — YoY Growth % by Country
# Using company_adaptation AI Integration rows as proxy
ai_yoy = df_comp[df_comp['adaptation_type'] == 'AI Integration'].groupby(
    ['country', 'year'])['digital_maturity_score'].mean().reset_index()
ai_yoy = ai_yoy.sort_values(['country', 'year'])
ai_yoy['YoY_Growth_Pct'] = ai_yoy.groupby('country')['digital_maturity_score'].pct_change().mul(100).round(2)
kpi3 = ai_yoy.dropna().groupby('country')['YoY_Growth_Pct'].mean().round(2).reset_index()
kpi3.columns = ['Country', 'Avg_YoY_AI_Velocity_Pct']
kpi3 = kpi3.sort_values('Avg_YoY_AI_Velocity_Pct', ascending=False)
print("\nKPI 3 — AI Adoption Velocity (Avg YoY Growth %) by Country:")
print(kpi3.to_string(index=False))

# KPI 4: Total AI Investment by Country
kpi4 = df_ai.groupby('country')['investment_usd'].sum().round(2).reset_index()
kpi4.columns = ['Country', 'Total_Investment_USD']
kpi4 = kpi4.sort_values('Total_Investment_USD', ascending=False)
print("\nKPI 4 — Total AI Investment by Country (USD):")
print(kpi4.to_string(index=False))

# KPI 5: Avg AI Error Rate % by Country
kpi5 = df_ai.groupby('country')['ai_error_rate_pct'].mean().round(2).reset_index()
kpi5.columns = ['Country', 'Avg_AI_Error_Rate_Pct']
kpi5 = kpi5.sort_values('Avg_AI_Error_Rate_Pct')
print("\nKPI 5 — Avg AI Error Rate % by Country:")
print(kpi5.to_string(index=False))

# KPI 6: Avg Productivity Gain % by AI Tool
kpi6 = df_ai.groupby('ai_tool')['productivity_gain_pct'].mean().round(2).reset_index()
kpi6.columns = ['AI_Tool', 'Avg_Productivity_Gain_Pct']
kpi6 = kpi6.sort_values('Avg_Productivity_Gain_Pct', ascending=False)
print("\nKPI 6 — Avg Productivity Gain % by AI Tool:")
print(kpi6.to_string(index=False))

# KPI 7: Jobs Displaced vs Jobs Created by Country
kpi7 = df_ai.groupby('country')[['jobs_displaced', 'jobs_created']].sum().round(0).reset_index()
kpi7['Net_Jobs'] = kpi7['jobs_created'] - kpi7['jobs_displaced']
kpi7.columns = ['Country', 'Jobs_Displaced', 'Jobs_Created', 'Net_Jobs']
kpi7 = kpi7.sort_values('Net_Jobs', ascending=False)
print("\nKPI 7 — Jobs Displaced vs Created by Country:")
print(kpi7.to_string(index=False))
```

```python
# ════════════════════════════════════════════════════════
# PAGE 2 — MEDIA CONSUMPTION KPIs
# Tables: df_media + df_app
# ════════════════════════════════════════════════════════
```

```python
print("=" * 55)
print("PAGE 2 — MEDIA CONSUMPTION KPIs")
print("=" * 55)

# KPI 1: Avg Daily Usage Hours by Platform
kpi1 = df_media.groupby('platform')['daily_usage_hrs'].mean().round(2).reset_index()
kpi1.columns = ['Platform', 'Avg_Daily_Usage_Hrs']
kpi1 = kpi1.sort_values('Avg_Daily_Usage_Hrs', ascending=False)
print("\nKPI 1 — Avg Daily Usage Hours by Platform:")
print(kpi1.to_string(index=False))

# KPI 2: Top Platforms by Country
kpi2 = df_media.groupby(['country', 'platform'])['daily_usage_hrs'].mean().round(2).reset_index()
kpi2.columns = ['Country', 'Platform', 'Avg_Daily_Usage_Hrs']
kpi2 = kpi2.sort_values(['Country', 'Avg_Daily_Usage_Hrs'], ascending=[True, False])
kpi2 = kpi2.groupby('Country').head(3)
print("\nKPI 2 — Top 3 Platforms by Country:")
print(kpi2.to_string(index=False))

# KPI 3: Media Type Share % by Decade
df_media['decade'] = (df_media['year'] // 10 * 10).astype(str) + 's'
kpi3 = df_media.groupby(['decade', 'media_type'])['daily_usage_hrs'].sum().reset_index()
kpi3['total'] = kpi3.groupby('decade')['daily_usage_hrs'].transform('sum')
kpi3['Share_Pct'] = (kpi3['daily_usage_hrs'] / kpi3['total'] * 100).round(2)
kpi3 = kpi3[['decade', 'media_type', 'Share_Pct']].sort_values(['decade', 'Share_Pct'], ascending=[True, False])
print("\nKPI 3 — Media Type Share % by Decade:")
print(kpi3.to_string(index=False))

# KPI 4: App Download Count by Platform
kpi4 = df_app.groupby('platform')['downloads_millions'].sum().round(2).reset_index()
kpi4.columns = ['Platform', 'Total_Downloads_Millions']
kpi4 = kpi4.sort_values('Total_Downloads_Millions', ascending=False)
print("\nKPI 4 — Total App Downloads (M) by Platform:")
print(kpi4.to_string(index=False))

# KPI 5: Ad Spend by Platform by Country
kpi5 = df_media.groupby(['country', 'platform'])['ad_spend_usd'].sum().round(2).reset_index()
kpi5.columns = ['Country', 'Platform', 'Total_Ad_Spend_USD']
kpi5 = kpi5.sort_values(['Country', 'Total_Ad_Spend_USD'], ascending=[True, False])
kpi5 = kpi5.groupby('Country').head(3)
print("\nKPI 5 — Top 3 Platforms by Ad Spend per Country:")
print(kpi5.to_string(index=False))

# KPI 6: Subscriber Growth Rate % by Platform
kpi6 = df_media.groupby(['platform', 'year'])['subscribers_millions'].mean().reset_index()
kpi6 = kpi6.sort_values(['platform', 'year'])
kpi6['YoY_Growth_Pct'] = kpi6.groupby('platform')['subscribers_millions'].pct_change().mul(100).round(2)
kpi6 = kpi6.dropna().groupby('platform')['YoY_Growth_Pct'].mean().round(2).reset_index()
kpi6.columns = ['Platform', 'Avg_Subscriber_Growth_Pct']
kpi6 = kpi6.sort_values('Avg_Subscriber_Growth_Pct', ascending=False)
print("\nKPI 6 — Avg Subscriber Growth Rate % by Platform:")
print(kpi6.to_string(index=False))

# KPI 7: Avg Penetration Rate % by Country
kpi7 = df_media.groupby('country')['penetration_rate_pct'].mean().round(2).reset_index()
kpi7.columns = ['Country', 'Avg_Penetration_Rate_Pct']
kpi7 = kpi7.sort_values('Avg_Penetration_Rate_Pct', ascending=False)
print("\nKPI 7 — Avg Penetration Rate % by Country:")
print(kpi7.to_string(index=False))
```

```python
# ════════════════════════════════════════════════════════
# PAGE 3 — DIGITAL TRANSFORMATION KPIs
# Tables: df_comp + df_ai
# ════════════════════════════════════════════════════════
```

```python
print("=" * 55)
print("PAGE 3 — DIGITAL TRANSFORMATION KPIs")
print("=" * 55)

# KPI 1: Avg Digital Maturity Score by Industry
kpi1 = df_comp.groupby('industry')['digital_maturity_score'].mean().round(2).reset_index()
kpi1.columns = ['Industry', 'Avg_Digital_Maturity_Score']
kpi1 = kpi1.sort_values('Avg_Digital_Maturity_Score', ascending=False)
print("\nKPI 1 — Avg Digital Maturity Score by Industry:")
print(kpi1.to_string(index=False))

# KPI 2: Avg Revenue Growth % by Industry
kpi2 = df_comp.groupby('industry')['revenue_growth_pct'].mean().round(2).reset_index()
kpi2.columns = ['Industry', 'Avg_Revenue_Growth_Pct']
kpi2 = kpi2.sort_values('Avg_Revenue_Growth_Pct', ascending=False)
print("\nKPI 2 — Avg Revenue Growth % by Industry:")
print(kpi2.to_string(index=False))

# KPI 3: AI Integration Rate % by Industry
kpi3 = df_comp[df_comp['adaptation_type'] == 'AI Integration'].groupby(
    'industry').size().reset_index(name='AI_Integration_Count')
total = df_comp.groupby('industry').size().reset_index(name='Total_Count')
kpi3 = kpi3.merge(total, on='industry')
kpi3['AI_Integration_Rate_Pct'] = (kpi3['AI_Integration_Count'] / kpi3['Total_Count'] * 100).round(2)
kpi3 = kpi3[['industry', 'AI_Integration_Rate_Pct']].sort_values('AI_Integration_Rate_Pct', ascending=False)
kpi3.columns = ['Industry', 'AI_Integration_Rate_Pct']
print("\nKPI 3 — AI Integration Rate % by Industry:")
print(kpi3.to_string(index=False))

# KPI 4: Leading Country by Avg Digital Maturity
kpi4 = df_comp.groupby('country')['digital_maturity_score'].mean().round(2).reset_index()
kpi4.columns = ['Country', 'Avg_Digital_Maturity_Score']
kpi4 = kpi4.sort_values('Avg_Digital_Maturity_Score', ascending=False)
print("\nKPI 4 — Leading Country by Avg Digital Maturity Score:")
print(kpi4.to_string(index=False))

# KPI 5: Top Industries by AI Adoption
kpi5 = df_ai.groupby('sector')['ai_adoption_rate_pct'].mean().round(2).reset_index()
kpi5.columns = ['Sector', 'Avg_AI_Adoption_Rate_Pct']
kpi5 = kpi5.sort_values('Avg_AI_Adoption_Rate_Pct', ascending=False)
print("\nKPI 5 — Top Industries by AI Adoption Rate %:")
print(kpi5.to_string(index=False))

# KPI 6: Tech Investment vs Digital Maturity Correlation
corr = df_comp[['tech_investment_usd', 'digital_maturity_score']].corr().round(4)
print("\nKPI 6 — Tech Investment vs Digital Maturity Correlation:")
print(corr.to_string())

# KPI 7: Avg Digital Revenue % by Industry
kpi7 = df_comp.groupby('industry')['digital_revenue_pct'].mean().round(2).reset_index()
kpi7.columns = ['Industry', 'Avg_Digital_Revenue_Pct']
kpi7 = kpi7.sort_values('Avg_Digital_Revenue_Pct', ascending=False)
print("\nKPI 7 — Avg Digital Revenue % by Industry:")
print(kpi7.to_string(index=False))
```

```python
# ════════════════════════════════════════════════════════
# PAGE 4 — COUNTRY COMPETITIVENESS KPIs
# Tables: All 4
# ════════════════════════════════════════════════════════
```

```python
print("=" * 55)
print("PAGE 4 — COUNTRY COMPETITIVENESS KPIs")
print("=" * 55)

# KPI 1: AI Readiness by Country
# Composite: Avg AI Adoption Rate + Avg Productivity Gain - Avg Error Rate
kpi1 = df_ai.groupby('country').agg(
    Avg_Adoption_Rate  = ('ai_adoption_rate_pct', 'mean'),
    Avg_Productivity   = ('productivity_gain_pct', 'mean'),
    Avg_Error_Rate     = ('ai_error_rate_pct', 'mean')
).round(2).reset_index()
kpi1['AI_Readiness_Score'] = (
    kpi1['Avg_Adoption_Rate'] * 0.5 +
    kpi1['Avg_Productivity'] * 0.3 -
    kpi1['Avg_Error_Rate'] * 0.2
).round(2)
kpi1 = kpi1.sort_values('AI_Readiness_Score', ascending=False)
print("\nKPI 1 — AI Readiness Score by Country:")
print(kpi1.to_string(index=False))

# KPI 2: Digital Maturity by Country
kpi2 = df_comp.groupby('country')['digital_maturity_score'].mean().round(2).reset_index()
kpi2.columns = ['Country', 'Avg_Digital_Maturity_Score']
kpi2 = kpi2.sort_values('Avg_Digital_Maturity_Score', ascending=False)
print("\nKPI 2 — Avg Digital Maturity Score by Country:")
print(kpi2.to_string(index=False))

# KPI 3: Media Shift by Country
# Streaming + Social Media share as % of total usage
df_media['is_digital'] = df_media['media_type'].isin(['Streaming', 'Social Media', 'Internet'])
kpi3 = df_media.groupby('country').agg(
    Digital_Hours = ('daily_usage_hrs', lambda x: x[df_media.loc[x.index, 'is_digital']].sum()),
    Total_Hours   = ('daily_usage_hrs', 'sum')
).reset_index()
kpi3['Digital_Media_Share_Pct'] = (kpi3['Digital_Hours'] / kpi3['Total_Hours'] * 100).round(2)
kpi3 = kpi3[['country', 'Digital_Media_Share_Pct']].sort_values('Digital_Media_Share_Pct', ascending=False)
kpi3.columns = ['Country', 'Digital_Media_Share_Pct']
print("\nKPI 3 — Digital Media Shift % by Country:")
print(kpi3.to_string(index=False))

# KPI 4: App Economy by Country
kpi4 = df_app.groupby('country').agg(
    Total_Downloads_M  = ('downloads_millions', 'sum'),
    Total_Revenue_USD  = ('revenue_usd', 'sum'),
    Avg_Rating         = ('avg_rating', 'mean')
).round(2).reset_index()
kpi4 = kpi4.sort_values('Total_Downloads_M', ascending=False)
print("\nKPI 4 — App Economy by Country:")
print(kpi4.to_string(index=False))

# KPI 5: Overall Competitiveness Score
# Composite of all 4 dimensions — normalised 0-100
from sklearn.preprocessing import MinMaxScaler

comp_df = kpi1[['country', 'AI_Readiness_Score']].merge(
    kpi2, left_on='country', right_on='Country').drop('Country', axis=1).merge(
    kpi3, left_on='country', right_on='Country').drop('Country', axis=1).merge(
    kpi4[['country', 'Total_Downloads_M']], on='country')

scaler = MinMaxScaler(feature_range=(0, 100))
comp_df[['AI_Score_N', 'Maturity_N', 'Media_N', 'App_N']] = scaler.fit_transform(
    comp_df[['AI_Readiness_Score', 'Avg_Digital_Maturity_Score',
             'Digital_Media_Share_Pct', 'Total_Downloads_M']]
)
comp_df['Overall_Competitiveness_Score'] = (
    comp_df['AI_Score_N'] * 0.35 +
    comp_df['Maturity_N'] * 0.30 +
    comp_df['Media_N']   * 0.20 +
    comp_df['App_N']     * 0.15
).round(2)

kpi5 = comp_df[['country', 'Overall_Competitiveness_Score']].sort_values(
    'Overall_Competitiveness_Score', ascending=False)
kpi5.columns = ['Country', 'Overall_Competitiveness_Score']
print("\nKPI 5 — Overall Competitiveness Score by Country (0-100):")
print(kpi5.to_string(index=False))

# KPI 6: Total Tech Investment by Country (cross-table)
kpi6_ai   = df_ai.groupby('country')['investment_usd'].sum().reset_index()
kpi6_comp = df_comp.groupby('country')['tech_investment_usd'].sum().reset_index()
kpi6 = kpi6_ai.merge(kpi6_comp, on='country')
kpi6['Total_Investment_USD'] = kpi6['investment_usd'] + kpi6['tech_investment_usd']
kpi6 = kpi6[['country', 'Total_Investment_USD']].sort_values('Total_Investment_USD', ascending=False)
kpi6.columns = ['Country', 'Total_Investment_USD']
print("\nKPI 6 — Total Combined Investment by Country (USD):")
print(kpi6.to_string(index=False))
```
